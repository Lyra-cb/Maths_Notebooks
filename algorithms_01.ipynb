```python
import time

# Create a list and dictionary of 1 million patients
n = 1_000_000
patient_list = list(range(n))
patient_dict = {i: True for i in range(n)}

target = 999_999  # worst case — last item!

# O(n) — linear search through list
start = time.time()
result = target in patient_list
list_time = time.time() - start

# O(1) — hash table lookup
start = time.time()
result = target in patient_dict
dict_time = time.time() - start

print(f"List search O(n):  {list_time:.6f} seconds")
print(f"Dict lookup O(1):  {dict_time:.6f} seconds")
print(f"Dictionary is {list_time/dict_time:.0f}x faster!")
```

```python
import matplotlib.pyplot as plt
import numpy as np

n = np.array([1, 10, 100, 1000, 10000, 100000, 1000000])

o1 = np.ones(len(n))          # O(1) — constant
ologn = np.log2(n)            # O(log n) — logarithmic  
on = n                         # O(n) — linear
on2 = n**2                     # O(n²) — quadratic

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# Left chart — all four (hard to see O(1) and O(log n)!)
ax1.plot(n, o1, label='O(1) — constant', linewidth=2)
ax1.plot(n, ologn, label='O(log n) — logarithmic', linewidth=2)
ax1.plot(n, on, label='O(n) — linear', linewidth=2)
ax1.plot(n, on2, label='O(n²) — quadratic', linewidth=2)
ax1.set_xlabel('Number of records (n)')
ax1.set_ylabel('Steps required')
ax1.set_title('All complexities')
ax1.legend()

# Right chart — zoom in, hiding O(n²) so we can see the others
ax2.plot(n, o1, label='O(1) — constant', linewidth=2)
ax2.plot(n, ologn, label='O(log n) — logarithmic', linewidth=2)
ax2.plot(n, on, label='O(n) — linear', linewidth=2)
ax2.set_ylim(0, 1000)
ax2.set_xlabel('Number of records (n)')
ax2.set_ylabel('Steps required')
ax2.set_title('Zoomed in (excluding O(n²))')
ax2.legend()

plt.tight_layout()
plt.show()
```
